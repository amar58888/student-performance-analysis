```python
import pandas as pd 
df = pd.read_csv("data.csv")
df.head()
```

```python
df.describe()
```

```python
df['G1'].mean()
df['G2'].mean()
df['G3'].mean()
```

```python
print("G1 Average:", df['G1'].mean())
print("G2 Average:", df['G2'].mean())
print("G3 Average:", df['G3'].mean())
```

```python
df.groupby('sex')['G3'].mean()
```

```python
import matplotlib.pyplot as plt

df.groupby('sex')['G3'].mean().plot(kind='bar')
plt.title("Average Final Score by Gender")
plt.show()
```

```python
import matplotlib.pyplot as plt
df.groupby('sex')['G3'].mean().plot(kind='bar')
plt.title("Average Final Score by Gender")
plt.show()
```

## Insights
- The average score slightly decreases from G1 to G3.
- Male students scored higher than female students in final grades.
- Most students scored around 10 marks.
